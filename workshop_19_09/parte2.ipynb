## Verificando a qualidade dos dados

```python
# Verificando se há duplicatas

df.duplicated().sum()
```

```python
# Se houver dados duplicados, verificamos quais são:

df[df.duplicated()]
```

**Conclusão 1:** não há duplicatas no dataset analisado.

```python
# Verificando se há valores faltantes

dados_faltantes_soma = df.isnull().sum()
dados_faltantes_porc = df.isnull().mean() * 100

df_dados_faltantes = pd.DataFrame({"Dados faltantes (soma)": dados_faltantes_soma, 
                                   "Dados faltantes (porcentagem)": dados_faltantes_porc})
df_dados_faltantes
```

**Conclusão 2:** Há valores faltantes em 2 colunas do dataset e ambas são pertinentes à pergunta de pesquisa. No entanto, só há 1 dado faltante em cada uma das referidas colunas o que equivale a 0.5% da coluna, ou seja, o dataset tem 200 linhas e apenas 1 delas não tem a quantidade de horas de sono do respectivo indivíduo e apenas 1 (pode ser a mesma ou não) não tem informação do nível de estresse. Logo, essas linhas sem dados são de pouca importância para a análise, poderíamos apagá-los para uma análise mais limpa do dataset. Faremos isso a seguir.

```python
df = df.dropna(subset=["horas_sono", "nivel_estresse"])
```

```python
# Verifica se ainda há valores faltantes nas colunas de interesse

dados_faltantes_soma = df.isnull().sum()
dados_faltantes_porc = df.isnull().mean() * 100

df_dados_faltantes = pd.DataFrame({"Dados faltantes (soma)": dados_faltantes_soma, 
                                   "Dados faltantes (porcentagem)": dados_faltantes_porc})
df_dados_faltantes
```

**Conclusão 3:** Na limpeza do dataset, portanto, apenas apaguei as linhas que não tinham dados para horas_sono ou nivel_estresse. Como já não tinha dados faltantes nas outras colunas, agora temos informações em **todas** as linhas e colunas do dataset.

```python
# corrigindo o caso de categorias iguais que estão como diferentes por causa do case sensitive

df["alimentacao"] = df["alimentacao"].str.lower()
df["alimentacao"] = df["alimentacao"].str.strip()

print(df["alimentacao"].unique())
```
