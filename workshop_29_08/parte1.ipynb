**Variável:** é um espaço de memória do computador para guardar um valor. Ela recebe um nome para que possamos usar esse valor depois. Variáveis ajudam a organizar informações dentro do programa. **O nome da variável deve ajudar a entender o que ela representa**.

Em Python, os valores guardados nas variáveis têm tipos diferentes. Os tipos básicos são números inteiros (**int**), números decimais (**float**), textos (**string**) e booleano (**bool**). Cada tipo é usado para representar um tipo específico de informação.

```python
# Variáveis

nome = "Larissa"
idade = 23
altura = 1.70
estudante = True

print("Nome:", nome)
print("Idade:", idade)
print("Altura:", altura)
print("É estudante?", estudante)
```

```python
# Tipo de uma variável (em python, usamos type())

print(type(nome))       # string (texto)                  - categórica
print(type(idade))      # int (inteiro)                   - numérica
print(type(altura))     # float (decimal)                 - numérica
print(type(estudante))  # bool (booleano - true or false) - categórica
```

Python pode ser usado como uma calculadora. É possível somar, subtrair, multiplicar e dividir valores, usando números ou variáveis do tipo int ou float. **Operações permitem transformar dados em informação**. Com elas podemos calcular médias, diferenças e totais. Elas são a base para análises mais complexas.

```python
# Operações aritméticas

a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 3)
```

```python
# Operações de comparação

salario_ana = 3000
salario_larissa = 2500

print(salario_ana == salario_larissa)
print(56 == "56")
print(30 != "30")
print(4 <= 4)
```

```python
# Operações lógicas

print(3 == 4 and 10 == 10)
print(salario_ana > salario_larissa or salario_ana > 5000)
```

**Bibliotecas:** conjuntos de códigos prontos. Elas evitam que precisemos escrever tudo do zero. Com bibliotecas, podemos fazer tarefas complexas com poucas linhas de código, economizando tempo e reduzindo erros. As bibliotecas são criadas e testadas por muitas pessoas, sendo amplamente usadas na programação profissiona. Vamos explorar **Pandas, Matplotlib e Seaborn**.

**Biblioteca Pandas:** usada para trabalhar com dados em tabelas. Ela permite ler arquivos como CSV e organizar informações, é muito usada em análise de dados.

```python
import pandas as pd

dados = {
    "nome": ["Ana", "Beatriz", "Carla"],
    "idade": [16, 17, 16],
    "nota": [7.5, 8.0, 6.8]
}

df = pd.DataFrame(dados)

print("Tabela de dados:")
print(df)

print("#######"*5)
print("Média das notas:")
print(df["nota"].mean())

print("#######"*5)
print("Menor nota: ", df["nota"].min())
print("A maior nota foi ", df["nota"].max(), "e é da", df.loc[df["nota"].idxmax(), "nome"])
```

**Matplotlib:** usada para criar gráficos em Python. Com ela, é possível visualizar dados de forma clara uma vez que gráficos ajudam a entender melhor as informações.

```python
# biblioteca para lidar com gráficos
import matplotlib.pyplot as plt

nomes = df["nome"]
notas = df["nota"]

plt.bar(nomes, notas, color="orange")
plt.title("Notas das estudantes")
plt.xlabel("Nome")
plt.ylabel("Nota")
plt.show()
```

**Seaborn:** biblioteca baseada em Matplotlib e muito usada para visualizações de dados também.

```python
import seaborn as sns

sns.barplot(x="nome", y="nota", data=df, color="orange")
plt.title("Notas das estudantes (Seaborn)")
plt.show()
```
