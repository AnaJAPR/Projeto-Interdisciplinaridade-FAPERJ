```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
dados = { "nome" :[ "Ana", "Beatriz", "Carla"],
         "idade" :[ 16, 18, 25],
         "salario":[ 10000, 8500, 15000]}
df = pd.DataFrame (dados)
```

```python
print(df)
```

```python
print(df["nome"])
```

```python
media_salario= df["salario"]. mean()
print(media_salario)
```

```python
maior_salario= df["salario"].max()
menor_idade= df["idade"].min()
idx_mais_velha= df["idade"].idxmax()
mais_velha= df["nome"][idx_mais_velha]

print(mais_velha)
```

```python
plt.bar (df["nome"], df["salario"], color="pink")
plt.title ("salario por profissional")
plt.xlabel ("nome")
plt.ylabel ("salario")
plt.show ()
```
